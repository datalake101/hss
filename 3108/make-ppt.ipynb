```python
from make_deck import parse, build, PALETTES
import glob
import re

# Convert Lecture 1
title, label, slides = parse('Lecture 01 Slides PlainText.txt')
build(title, label, slides, 'crimson', 'Lecture 01 Slides PlainText_premium.pptx')
print('Done!')
```

```python
# Or batch convert all lectures in the directory:
for f in sorted(glob.glob("[lL]ecture*.txt")):
    out = re.sub(r"\.txt$", "", f) + "_premium.pptx"
    t, l, s = parse(f)
    build(t, l, s, "crimson", out)
print("All lectures converted!")
```
